```python
import pandas as pd

# Carregar o arquivo
df = pd.read_csv('/content/Obesity.csv')

print(f"✅ Arquivo carregado com sucesso! {df.shape[0]} linhas e {df.shape[1]} colunas.")
df.head()
```

```python
# Renomear colunas para português
df = df.rename(columns={
    'Gender': 'Genero',
    'Age': 'Idade',
    'Height': 'Altura',
    'family_history': 'Histórico_Familiar',
    'FAVC': 'Alimentos_Caloricos',
    'FCVC': 'Come_Vegetais',
    'NCP': 'Qtd_Refeicoes',
    'CAEC': 'Come_Entre_Refeicao',
    'SMOKE': 'Fumante',
    'CH2O': 'Quanto_Agua',
    'SCC': 'Monitora_Calorias',
    'FAF': 'Freq_Exercicio',
    'TUE': 'Tempo_Tela',
    'CALC': 'Freq_Alcool',
    'MTRANS': 'Meio_Transporte',
    'Obesity': 'Nível_obesidade'
})

print("✅ Colunas renomeadas com sucesso!")
df.head()
```

```python
# ====================== TRADUÇÃO DOS DADOS ======================

# Dicionários de tradução
traducao_sim_nao = {'yes': 'Sim', 'no': 'Não'}

traducao_frequencia = {
    'no': 'Não',
    'Sometimes': 'Às vezes',
    'Frequently': 'Frequentemente',
    'Always': 'Sempre'
}

# Aplicando traduções
df['Genero'] = df['Genero'].map({'Female': 'Feminino', 'Male': 'Masculino'})

df['Histórico_Familiar'] = df['Histórico_Familiar'].map(traducao_sim_nao)
df['Alimentos_Caloricos'] = df['Alimentos_Caloricos'].map(traducao_sim_nao)
df['Fumante'] = df['Fumante'].map(traducao_sim_nao)
df['Monitora_Calorias'] = df['Monitora_Calorias'].map(traducao_sim_nao)

df['Come_Entre_Refeicao'] = df['Come_Entre_Refeicao'].map(traducao_frequencia)
df['Freq_Alcool'] = df['Freq_Alcool'].map(traducao_frequencia)

# Mapeamentos numéricos
df['Qtd_Refeicoes'] = df['Qtd_Refeicoes'].map({1: '1 refeição', 2: '2 refeições',
                                                 3: '3 refeições', 4: '4 refeições ou mais'})

df['Come_Vegetais'] = df['Come_Vegetais'].map({1: 'Raramente', 2: 'Às vezes', 3: 'Sempre'})
df['Quanto_Agua'] = df['Quanto_Agua'].map({1: '< 1 L/dia', 2: '1-2 L/dia', 3: '> 2 L/dia'})
df['Freq_Exercicio'] = df['Freq_Exercicio'].map({0: 'Nenhuma', 1: '1-2x/semana',
                                                  2: '3-4x/semana', 3: '5x ou mais'})
df['Tempo_Tela'] = df['Tempo_Tela'].map({0: '0-2 horas/dia', 1: '3-5 horas/dia', 2: 'Mais de 5 horas/dia'})

df['Meio_Transporte'] = df['Meio_Transporte'].map({
    'Public_Transportation': 'Transporte Público',
    'Automobile': 'Carro',
    'Motorbike': 'Moto',
    'Bike': 'Bicicleta',
    'Walking': 'A pé'
})

df['Nível_obesidade'] = df['Nível_obesidade'].map({
    'Insufficient_Weight': 'Abaixo do Peso',
    'Normal_Weight': 'Peso Normal',
    'Overweight_Level_I': 'Sobrepeso Nível I',
    'Overweight_Level_II': 'Sobrepeso Nível II',
    'Obesity_Type_I': 'Obesidade Tipo I',
    'Obesity_Type_II': 'Obesidade Tipo II',
    'Obesity_Type_III': 'Obesidade Tipo III'
})

# Arredondar valores numéricos
df['Idade'] = df['Idade'].round(2)
df['Altura'] = df['Altura'].round(2)

# Criar faixa etária
def definir_faixa_etaria(idade):
    if idade < 18:
        return 'Adolescente (<=18)'
    elif idade <= 29:
        return 'Jovem (18-29)'
    elif idade <= 44:
        return 'Adulto (30-44)'
    elif idade <= 59:
        return 'Meia-idade (45-59)'
    else:
        return 'Idoso (60+)'

df['Faixa_etaria'] = df['Idade'].apply(definir_faixa_etaria)

# Salvar versão traduzida
df.to_csv('Obesity_Traduzido_PowerBI.csv', index=False, encoding='utf-8')

print("✅ Tradução concluída com sucesso!")
print(f"Arquivo salvo: 'Obesity_Traduzido_PowerBI.csv'")
df.head()
```
